```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv(r"C:\Users\aziad\Downloads\Telco_Customer_Churn_Cleaned.csv")

# 1) جدول الأعداد مع الإجماليات
ct_counts = pd.crosstab(
    df["PaperlessBilling"], df["PaymentMethod"], margins=True
)
print("Counts:\n", ct_counts)

# 2) جدول النسب المئوية داخل كل فئة من PaperlessBilling
ct_pct = (
    pd.crosstab(df["PaperlessBilling"], df["PaymentMethod"], normalize="index")
    * 100
).round(2)
print("\nRow Percentages (%):\n", ct_pct)

# 3) رسم بياني لتوزيع PaymentMethod داخل كل فئة من PaperlessBilling
ax = ct_pct.plot(kind="bar", figsize=(9, 4.5), width=0.6)
plt.title("PaymentMethod Distribution within PaperlessBilling (%)")
plt.xlabel("PaperlessBilling")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.ylim(0, 52)
plt.legend(title="PaymentMethod", bbox_to_anchor=(1.02, 1), loc="upper left")

for container in ax.containers:
  ax.bar_label(container, fmt="%.1f%%", padding=2, fontsize=8)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

file_path = r"C:\Users\aziad\Downloads\Telco_Customer_Churn_Cleaned.csv"
df = pd.read_csv(file_path)

# 1) ملخص إحصائي للفاتورة الشهرية حسب نوع الإنترنت
summary = (
    df.groupby("InternetService")["MonthlyCharges"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .round(2)
)
print(summary)

# 2) رسم Boxplot لمقارنة توزيع MonthlyCharges بين فئات InternetService
plt.figure(figsize=(8, 4.5))
sns.boxplot(
    data=df,
    x="InternetService",
    y="MonthlyCharges",
    palette=["#DD8452", "#4C72B0", "#55A868"],
)
plt.title("MonthlyCharges Distribution by InternetService")
plt.xlabel("InternetService")
plt.ylabel("MonthlyCharges ($)")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

file_path = r"C:\Users\aziad\Downloads\Telco_Customer_Churn_Cleaned.csv"
df = pd.read_csv(file_path)

num_cols = ["tenure", "MonthlyCharges", "TotalCharges"]

# 1) ملخص إحصائي للأعمدة العددية
print("Summary Statistics:\n", df[num_cols].describe().round(2))

# 2) مصفوفة الارتباط الخطي (Pearson Correlation)
corr_matrix = df[num_cols].corr()
print("\nCorrelation Matrix:\n", corr_matrix.round(4))

# 3) رسومات Scatter لكل زوج من الأعمدة العددية
fig, axes = plt.subplots(1, 3, figsize=(15, 4.2))
pairs = [
    ("tenure", "TotalCharges", "#4C72B0"),
    ("MonthlyCharges", "TotalCharges", "#DD8452"),
    ("tenure", "MonthlyCharges", "#55A868"),
]

for ax, (x_col, y_col, color) in zip(axes, pairs):
  r_val = df[x_col].corr(df[y_col])
  sns.scatterplot(
      data=df, x=x_col, y=y_col, alpha=0.25, s=12, color=color, ax=ax
  )
  ax.set_title(f"{x_col} vs {y_col} (r = {r_val:.2f})")
  ax.set_xlabel(x_col)
  ax.set_ylabel(y_col)

plt.tight_layout()
plt.show()
```
